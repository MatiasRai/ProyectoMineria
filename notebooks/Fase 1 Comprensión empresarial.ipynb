# Objetivo Empresarial
- **Contexto del videojuego StarCraft II:** StarCraft II es un juego de estrategia en tiempo real desarrollado por Blizzard Entertainment, lanzado en 2010,el juego es conocido por su competitivo modo multijugador y su popularidad en los eSports,los jugadores seleccionan una de las tres facciones y deben gestionar recursos, realizar estrategias y ejecutar acciones rápidas para ganar.
- **Objetivo principal:** El objetivo principal de este análisis es predecir el nivel de habilidad de los jugadores de StarCraft II en función de sus comportamientos en el juego, como sus horas de juego, acciones por minuto (APM), y otras variables clave. A través de este análisis, el objetivo es ofrecer insights sobre las características que influyen en el rendimiento de los jugadores, con el fin de proporcionar recomendaciones tanto para los jugadores como para los desarrolladores del juego, ayudando a mejorar la experiencia de juego y la competitividad.

    Además, se busca optimizar las decisiones de diseño del juego para promover un entorno más equilibrado, con un enfoque particular en mejorar las ligas más bajas y fomentar la competitividad en todas las fases del juego.



# Alcance del proyecto  
El análisis se centrará en un conjunto de datos proveniente de jugadores de StarCraft II, con información detallada sobre sus acciones en el juego y su clasificación en ligas. Se utilizarán las siguientes variables clave del dataset:
    
**Variables demográficas:** edad del jugador

**Variables de Actividad en el Juego:** Horas jugadas semanalmente y total de horas, APM, unidades creadas, entre otras.

**Objetivo de Predicción:** Predecir el nivel de liga del jugador (LeagueIndex) basado en su actividad en el juego.

El proyecto no incluirá la simulación de nuevos datos ni la intervención directa sobre las estrategias del juego, sino un análisis descriptivo y predictivo basado en el dataset proporcionado. Se explorarán posibles correlaciones entre las variables, se realizará un análisis de patrones y se propondrán modelos predictivos utilizando técnicas de Machine Learning.
    

 



# Criterios de éxito
- **Precisión de los Modelos Predictivos:** La precisión del modelo de predicción del nivel de liga deberá ser superior al 70%. Se utilizarán métricas como la exactitud, precisión, recall y F1-score para evaluar los modelos de clasificación.
- **Identificación de Patrones Relevantes:** El proyecto debe identificar patrones significativos entre las variables que expliquen las diferencias en el nivel de habilidad de los jugadores. Esto podría incluir, por ejemplo, la relación entre las horas jugadas y las acciones por minuto.
- **Generación de Insights Prácticos:** A partir del análisis, deben generarse recomendaciones concretas tanto para los jugadores como para los desarrolladores del juego, basadas en los patrones y relaciones observadas.
- **Claridad y Calidad del Informe Final:** El informe final debe ser claro, técnico y visualmente accesible, con una interpretación sólida de los resultados y las visualizaciones adecuadas. Se deben usar gráficos que ayuden a comprender los insights obtenidos.


# Consideraciones Éticas y de Privacidad
Es importante abordar las consideraciones éticas y de privacidad, especialmente cuando se trabaja con datos de jugadores. Aquí te dejo un ejemplo de consideraciones éticas:
- **Privacidad de los Jugadores::** Se debe asegurar que los datos de los jugadores no contengan información personal identificable (PII). En este caso, como el dataset solo incluye variables como "edad" y "LeagueIndex", no parece haber un riesgo directo de exposición de la identidad de los jugadores. Sin embargo, es esencial asegurarse de que no se utilicen datos fuera de lo permitido en el acuerdo de privacidad de Blizzard Entertainment.
- **Uso de Datos en Competencias:** Dado que los datos provienen de un entorno de competencia, es crucial no utilizar estos datos para manipular o crear desigualdades en las competiciones (por ejemplo, al tratar de mejorar el rendimiento de ciertos jugadores de manera artificial).
- **Transparencia y Explicabilidad de los Modelos:** Los modelos predictivos empleados deben ser explicables. No se deben utilizar algoritmos de caja negra sin justificación adecuada, y cualquier predicción o recomendación generada debe estar basada en patrones que sean comprensibles y transparentes.
- **Consentimiento Informado:** Si los datos provinieran de una fuente más personal (por ejemplo, datos directos de los jugadores), se debería garantizar que los jugadores han otorgado su consentimiento para el uso de sus datos con fines analíticos y de investigación.
