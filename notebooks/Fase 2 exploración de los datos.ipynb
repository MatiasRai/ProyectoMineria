**Creamos un boxplot para visualizar los outliers de las variables numéricas**

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df[['APM', 'HoursPerWeek', 'TotalHours']])
plt.title("Boxplot de APM, HoursPerWeek, y TotalHours")
plt.show()
```

**Usaremos IQR para detecteras outliers, esto basicamente es para hacerlo de manera más estadistica.**

```python
# Asegúrate de que las columnas sean numéricas
df['APM'] = pd.to_numeric(df['APM'], errors='coerce')
df['HoursPerWeek'] = pd.to_numeric(df['HoursPerWeek'], errors='coerce')
df['TotalHours'] = pd.to_numeric(df['TotalHours'], errors='coerce')
```

```python
# Calcular el IQR (Rango Intercuartílico)
Q1 = df[['APM', 'HoursPerWeek', 'TotalHours']].quantile(0.25)
Q3 = df[['APM', 'HoursPerWeek', 'TotalHours']].quantile(0.75)
IQR = Q3 - Q1

# Definir los límites inferior y superior para detectar outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Ver cuántos valores están fuera de los límites
outliers = (df[['APM', 'HoursPerWeek', 'TotalHours']] < lower_bound) | (df[['APM', 'HoursPerWeek', 'TotalHours']] > upper_bound)
print(outliers.sum())
```

**Exploraremos la distribución de las variables numéricas con histogramas.**

```python
# Visualizar la distribución de algunas variables numéricas
plt.figure(figsize=(12, 6))

# APM
plt.subplot(1, 3, 1)
sns.histplot(df['APM'], kde=True, color='blue')
plt.title('Distribución de APM')

# HoursPerWeek
plt.subplot(1, 3, 2)
sns.histplot(df['HoursPerWeek'], kde=True, color='green')
plt.title('Distribución de HoursPerWeek')

# TotalHours
plt.subplot(1, 3, 3)
sns.histplot(df['TotalHours'], kde=True, color='red')
plt.title('Distribución de TotalHours')

plt.tight_layout()
plt.show()
```

**Haremos una matriz de correlación**

```python
# Convertiremostodas las columnas del DataFrame a tipo numérico
df = df.apply(pd.to_numeric, errors='coerce')

# Verificar la conversión
print(df.dtypes)
```

```python
# Verificaremos si hay valores nulos
print(df.isnull().sum())
```

```python
# Calcularemos la matriz de correlación
corr = df.corr()

# Crear un heatmap
plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Matriz de Correlación')
plt.show()
```

**Veremos mediante un gráfico la relación entre APM y HoursPerWeek.**

```python
# Crear un gráfico de dispersión para ver la relación entre APM y HoursPerWeek
plt.figure(figsize=(10,6))
sns.scatterplot(x='HoursPerWeek', y='APM', data=df)
plt.title("Relación entre HoursPerWeek y APM")
plt.show()
```

**Calculamos la correlación entre HoursPerWeek y APM.**

```python
correlation = df['HoursPerWeek'].corr(df['APM'])
print(f"Correlación entre HoursPerWeek y APM: {correlation}")
```

*El resultado de la correlación nos indica que no hay una relación directa entre estas dos variables.*
